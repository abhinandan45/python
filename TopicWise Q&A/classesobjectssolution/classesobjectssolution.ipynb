Create a class named `BankAccount` with private attributes `account_number` and `balance`. Add methods to deposit and withdraw money, and to check the balance. Create an object of the class and perform some operations.

````python
class BankAccount:
    def __init__(self, account_number, balance=0):
        self.__account_number = account_number
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient balance!")
        else:
            self.__balance -= amount

    def check_balance(self):
        return self.__balance

# Test
# account = BankAccount('12345678', 1000)
# account.deposit(500)
# account.withdraw(200)
# print(account.check_balance())  # 1300
# account.withdraw(2000)  # Insufficient balance!
````

### Assignment 8: Abstract Base Class

Create an abstract base class named `Shape` with an abstract method `area`. Create derived classes `Circle` and `Square` that implement the `area` method. Create objects of the derived classes and call the `area` method.

````python
from abc import ABC, abstractmethod
import math 

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

# Test
# circle = Circle(5)
# square = Square(4)
# print(circle.area())  # 78.53981633974483
# print(square.area())  # 16
````

### **Essential Questions on Basics**

1. **What is a class, and how is it different from an object?**  
   - A **class** is a blueprint or template for creating objects. It defines the attributes (data) and methods (functions) that the objects of the class will have.  
   - An **object** is an instance of a class. It is a concrete realization of the class.  
   **Example:**
   ```python
   class Car:
       def __init__(self, brand, model):
           self.brand = brand
           self.model = model

   car1 = Car("Toyota", "Camry")  # Object (instance) of class Car
   ```

2. **What is the purpose of the `__init__()` method? Is it mandatory to define it in every class?**  
   - The `__init__()` method is a special method (constructor) used to initialize an object’s attributes when the object is created.
   - It is **not mandatory** to define `__init__()` in every class. If not defined, Python provides a default constructor.
   **Example:**
   ```python
   class Person:
       def __init__(self, name, age):
           self.name = name
           self.age = age

   person = Person("Alice", 25)
   ```

3. **Why is `self` used in Python classes? Can you rename it to something else?**  
   - `self` represents the instance of the class. It is used to access instance attributes and methods.  
   - Yes, you can rename `self` to any other valid variable name, but it is a convention to use `self`.  
   **Example:**
   ```python
   class MyClass:
       def __init__(instance, value):
           instance.value = value
   obj = MyClass(10)
   print(obj.value)
   ```

4. **Explain how attributes (both instance and class attributes) work in Python.**  
   - **Instance Attributes:** Unique to each object, defined in the `__init__()` method using `self`.  
   - **Class Attributes:** Shared across all instances, defined directly within the class body.  
   **Example:**
   ```python
   class MyClass:
       class_attribute = "Shared"

       def __init__(self, value):
           self.instance_attribute = value

   obj1 = MyClass(10)
   obj2 = MyClass(20)
   print(MyClass.class_attribute)  # Shared
   print(obj1.instance_attribute)  # 10
   print(obj2.instance_attribute)  # 20
   ```

5. **How do you define and call instance methods in a class? Provide an example.**  
   - Instance methods are defined using `def` and take `self` as the first parameter. They are called using an object of the class.  
   **Example:**
   ```python
   class MyClass:
       def greet(self, name):
           return f"Hello, {name}!"

   obj = MyClass()
   print(obj.greet("Alice"))  # Output: Hello, Alice!
   ```

### **Frequently Asked Interview Questions**

6. **How does Python implement encapsulation? Can private attributes be accessed outside the class?**  
   - Python implements encapsulation by allowing attributes and methods to be marked as **public**, **protected** (prefix `_`), or **private** (prefix `__`).  
   - Private attributes can be accessed indirectly using getter/setter methods or directly with `_ClassName__attribute`.  
   **Example:**
   ```python
   class MyClass:
       def __init__(self):
           self.__private = "Hidden"

       def get_private(self):
           return self.__private

   obj = MyClass()
   print(obj.get_private())  # Access via method
   # print(obj.__private)    # AttributeError
   print(obj._MyClass__private)  # Access via name mangling
   ```

7. **What is method overloading, and how is it implemented in Python (if supported)?**  
   - **Method overloading** is having multiple methods with the same name but different parameters. Python does not natively support it; the last defined method overrides earlier ones.  
   - Achieved using default arguments or `*args` and `**kwargs`.  
   **Example:**
   ```python
   class MyClass:
       def greet(self, name=None):
           if name:
               return f"Hello, {name}!"
           return "Hello!"
   obj = MyClass()
   print(obj.greet())         # Hello!
   print(obj.greet("Alice"))  # Hello, Alice!
   ```

8. **Explain method overriding in Python with an example of inheritance.**  
   - **Method overriding** occurs when a subclass provides a specific implementation of a method defined in its parent class.  
   **Example:**
   ```python
   class Parent:
       def greet(self):
           return "Hello from Parent"

   class Child(Parent):
       def greet(self):
           return "Hello from Child"

   obj = Child()
   print(obj.greet())  # Output: Hello from Child
   ```

   obj1 = Singleton()
   obj2 = Singleton()

#### **14. Class to Manage a Team of Employees**
**Implementation:**  
```python
class EmployeeTeam:
    def __init__(self):
        self.employees = []

    def add_employee(self, name, age):
        self.employees.append({"name": name, "age": age})

    def remove_employee(self, name):
        self.employees = [emp for emp in self.employees if emp["name"] != name]

    def calculate_average_age(self):
        if not self.employees:
            return 0
        total_age = sum(emp["age"] for emp in self.employees)
        return total_age / len(self.employees)

# Example Usage
team = EmployeeTeam()
team.add_employee("Alice", 30)
team.add_employee("Bob", 25)
team.add_employee("Charlie", 35)
print("Average Age:", team.calculate_average_age())  # Output: 30.0
team.remove_employee("Bob")
print("Employees:", team.employees)  # Output: [{'name': 'Alice', 'age': 30}, {'name': 'Charlie', 'age': 35}]
```

#### **16. Online Shopping System Class**
**Implementation:**  
```python
class OnlineShopping:
    def __init__(self, tax_rate=0.1):
        self.orders = []
        self.tax_rate = tax_rate

    def add_order(self, item, price, quantity):
        self.orders.append({"item": item, "price": price, "quantity": quantity})

    def calculate_total_price(self):
        subtotal = sum(order["price"] * order["quantity"] for order in self.orders)
        tax = subtotal * self.tax_rate
        return subtotal + tax

    def display_orders(self):
        return self.orders

# Example Usage
shop = OnlineShopping(tax_rate=0.08)
shop.add_order("Laptop", 1000, 1)
shop.add_order("Mouse", 50, 2)
print("Orders:", shop.display_orders())
print("Total Price (with tax):", shop.calculate_total_price())  # Output: 1180.0
```

#### **17. Advantages of Using Classes Over Functions**
1. **Encapsulation:** Classes bundle data (attributes) and related behavior (methods) together, improving modularity and preventing data leakage.  
   - Example: An `Employee` class encapsulates employee details and associated methods.

2. **Reusability:** With classes, you can reuse and extend code using inheritance, reducing redundancy.  
   - Example: Create a parent class `Vehicle` and reuse it in `Car` and `Bike`.

3. **Abstraction:** Classes hide implementation details while exposing a clean interface for the user.  
   - Example: A `BankAccount` class provides deposit and withdraw methods without exposing internal computations.

4. **Scalability:** Classes are more suited for large applications where managing functions becomes difficult.  
   - Example: A shopping system class can track orders, manage inventory, and calculate payments.

5. **Polymorphism:** Classes allow you to define methods that behave differently based on the object using them, enabling flexibility.  
   - Example: A `draw()` method in different shapes (`Circle`, `Rectangle`) behaves differently.

6. **Ease of Maintenance:** Since data and methods are grouped logically in classes, they are easier to maintain and debug.
